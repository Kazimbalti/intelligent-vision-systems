In a notebook we show images with **matplotlib** instead of `cv2.imshow`, so we must convert **BGR → RGB** first.

```python
import sys; sys.path.insert(0, '..')
import cv2
import numpy as np
import matplotlib.pyplot as plt
from labutils import data
print('OpenCV', cv2.__version__, '| NumPy', np.__version__)
```

## 1. Load an image and show it correctly

```python
img = cv2.imread(data('scene.jpg'))
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].imshow(img); ax[0].set_title('wrong: BGR shown as RGB')
ax[1].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); ax[1].set_title('right: converted to RGB')
for a in ax: a.axis('off')
plt.show()
```

## 2. An image is just numbers

```python
print(img.shape, img.dtype)
print('pixel [100, 200] =', img[100, 200])
print(img[195:200, 315:320, 2])   # red channel of a 5x5 patch inside the sun
```

## 3. Histogram of brightness
The histogram shows how many pixels have each grey level. Otsu's method picks the threshold that best separates its two peaks.

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
T, bw = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
plt.figure(figsize=(10, 3))
plt.hist(gray.ravel(), bins=256, range=(0, 256), color='gray')
plt.axvline(T, color='red', label=f'Otsu T = {T:.0f}')
plt.legend(); plt.xlabel('grey level'); plt.ylabel('pixels'); plt.show()
plt.imshow(bw, cmap='gray'); plt.axis('off'); plt.show()
```
